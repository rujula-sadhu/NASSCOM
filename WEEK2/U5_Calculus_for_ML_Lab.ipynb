```python
# Core imports for the whole lab
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')      # symbolic variables we'll reuse
sp.init_printing()            # pretty-print symbolic math
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)

```

```python
# -----------------------------------------------------------
# 🔹 1A. NUMERICAL DERIVATIVE (finite difference)
# -----------------------------------------------------------

# The derivative is the slope: how much f changes for a tiny step h
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
# -----------------------------------------------------------
# 🔹 1B. SYMBOLIC DERIVATIVE (SymPy)
# -----------------------------------------------------------

expr = x**2
deriv = sp.diff(expr, x)          # differentiate w.r.t. x
print('d/dx (x**2) =', deriv)     # 2*x

# Evaluate the symbolic derivative at x = 3
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
import sympy as sp

# Define our original mathematical function: f(x) = x^3 + 2x
def g(x):
    return x**3 + 2*x

# --- STEP 1: NUMERICAL DERIVATIVE ---
# This estimates the slope using a tiny change in x (called 'h')
x_value = 2
h = 0.000001  # This is 1e-6 written out fully

# Formula for slope: (change in y) / (change in x)
y1 = g(x_value)          # y value at exactly x = 2
y2 = g(x_value + h)      # y value just a tiny bit past x = 2

numerical_slope = (y2 - y1) / h


# --- STEP 2: SYMBOLIC DERIVATIVE ---
# This finds the exact calculus rule formula, just like doing math on paper
x_symbol = sp.Symbol('x')  # Tell Python 'x' is an algebraic variable

# Find the derivative formula of g(x)
symbolic_formula = sp.diff(g(x_symbol), x_symbol)


# --- STEP 3: EVALUATE AND COMPARE ---
# Plug x = 2 into our new exact formula to find the perfect slope
exact_slope = symbolic_formula.subs(x_symbol, x_value)


# --- PRINT RESULTS ---
print("--- Step 1 ---")
print("Estimated (Numerical) Slope:", numerical_slope)

print("\n--- Step 2 ---")
print("Calculus Formula:", symbolic_formula)

print("\n--- Step 3 ---")
print("Exact (Symbolic) Slope:", exact_slope)
```

```python
# -----------------------------------------------------------
# 🔹 2A. PARTIAL DERIVATIVES
# -----------------------------------------------------------

f2 = x**2 + 3*x*y + y**2

# A partial derivative differentiates ONE variable, holding others fixed
print('df/dx =', sp.diff(f2, x))     # 2*x + 3*y
print('df/dy =', sp.diff(f2, y))     # 3*x + 2*y
```

```python
# -----------------------------------------------------------
# 🔹 2B. THE GRADIENT (vector of partials)
# -----------------------------------------------------------

# The gradient stacks every partial derivative into one vector
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)

# Evaluate the gradient at the point (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)   # points in the steepest-ascent direction
```

```python
import sympy as sp

# Define our symbols
x, y = sp.symbols('x y')

# Define the function h2
h2 = x**2 * y + sp.sin(y)

# --- Compute dh/dx and dh/dy ---
# sp.diff(function, variable) treats all other variables as constants
dh_dx = sp.diff(h2, x)  # Derivative with respect to x
dh_dy = sp.diff(h2, y)  # Derivative with respect to y


# --- Assemble the gradient list ---
# A gradient is just a list of all your partial derivatives
gradient_expr = [dh_dx, dh_dy]


# --- Evaluate at (x=2, y=0) ---
# We pass a dictionary to .subs() to replace both x and y at the same time
# Then we convert the results to floats or ints for clean numbers
point = {x: 2, y: 0}
gradient_at_point = [expr.subs(point) for expr in gradient_expr]


# --- PRINT RESULTS ---
print("--- Partial Derivatives ---")
print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

print("\n--- Gradient Formula ---")
print("Gradient =", gradient_expr)

print("\n--- Value at (x=2, y=0) ---")
print("Evaluated Gradient =", gradient_at_point)
```

```python

# -----------------------------------------------------------
# 🔹 3A. CHAIN RULE BY HAND vs SymPy
# -----------------------------------------------------------

# y = sin(x**2) is a composition: outer = sin(u), inner = u = x**2
# Chain rule:  dy/dx = cos(u) * du/dx = cos(x**2) * 2x
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
# -----------------------------------------------------------
# 🔹 3B. CHAINING THREE FUNCTIONS
# -----------------------------------------------------------

# y = (3x + 1)**4  -> outer^4, inner (3x+1)
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))   # 12*(3x+1)^3
```

```python
import sympy as sp

x = sp.symbols('x')
y = sp.exp(x**2 + 1)

# 1. By hand (as a SymPy expression)
by_hand = sp.exp(x**2 + 1) * (2 * x)

# 2. With sp.diff
by_sympy = sp.diff(y, x)

# 3. Confirm they match
difference = sp.simplify(by_hand - by_sympy)
is_matching = (difference == 0)

print(f"By Hand: {by_hand}")
print(f"By SymPy: {by_sympy}")
print(f"Match: {is_matching}")
```

```python
# -----------------------------------------------------------
# 🔹 4A. FORWARD PASS
# -----------------------------------------------------------

# Tiny toy problem: 4 samples, 3 input features, 5 hidden units, 1 output
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
# -----------------------------------------------------------
# 🔹 4B. BACKWARD PASS (gradients via the chain rule)
# -----------------------------------------------------------

# Work backwards from the loss, one link at a time
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
# -----------------------------------------------------------
# 🔹 4C. ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS
# -----------------------------------------------------------

lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

# Seed for reproducibility
np.random.seed(42)

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output

# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss
z1 = np.dot(Xb, Wa)
h = np.maximum(0, z1)
y_hat = np.dot(h, Wb)
loss_before = np.mean((y_hat - Yb) ** 2)

# 2. Backward pass: dy, dWb, dh, dz1, dWa
m = Xb.shape[0]
dy = 2 * (y_hat - Yb) / m
dWb = np.dot(h.T, dy)
dh = np.dot(dy, Wb.T)
dz1 = dh * (z1 > 0)
dWa = np.dot(Xb.T, dz1)

# 3. One step with lr = 0.05; print loss before and after
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb

# Recalculate forward pass for updated loss
z1_new = np.dot(Xb, Wa)
h_new = np.maximum(0, z1_new)
y_hat_new = np.dot(h_new, Wb)
loss_after = np.mean((y_hat_new - Yb) ** 2)

print(f"Loss before step: {loss_before}")
print(f"Loss after step: {loss_after}")
```

```python
# -----------------------------------------------------------
# 🔹 5A. THE HESSIAN (matrix of second derivatives = curvature)
# -----------------------------------------------------------

f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))
print('Hessian of f:')
sp.pprint(H)        # [[2, 3], [3, 2]]
```

```python
# -----------------------------------------------------------
# 🔹 5B. GRADIENT DESCENT ON A SIMPLE FUNCTION
# -----------------------------------------------------------

# Minimise f(x) = (x - 4)**2, whose minimum is at x = 4.
# Update rule:  x <- x - lr * f'(x),  with f'(x) = 2*(x - 4)
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```
